### Retriever Model (2 points)

Our goal for today is to build a model that measures similarity between question and answer. In particular, it maps both question and answer into fixed-size vectors such that:

Our model is a pair of $V_q(q)$ and $V_a(a)$ - networks that turn phrases into vectors.

__Objective:__ Question vector $V_q(q)$ should be __closer__ to correct answer vectors $V_a(a^+)$ than to incorrect ones $V_a(a^-)$ .

Both vectorizers can be anything you wish. For starters, let's use a couple of dense layers on top of the pre-trained encoder.




````python
class Vectorizer(nn.Module):
    def __init__(self, hid_size=256, bert=None, tokenizer=None):
        """ A small feedforward network on top of pre-trained encoder. 2-3 layers should be enough """
        super().__init__()
        if bert is None:
            model_name = 'sentence-transformers/bert-base-nli-mean-tokens'
            self.bert = AutoModel.from_pretrained(model_name)
            self.tokenizer = AutoTokenizer.from_pretrained(model_name)
        else:
            self.bert = bert
            self.tokenizer = tokenizer
        self.hid_size = hid_size

        # define layers on top of BERT
        self.dropout = nn.Dropout(p=0.1)  # at least 10% dropout
        self.fc1 = nn.Linear(self.bert.config.hidden_size, hid_size)
        self.fc2 = nn.Linear(hid_size, hid_size)
        self.out = nn.Linear(hid_size, hid_size)  # final linear output

    def forward(self, input_phrases):
        # Tokenize input phrases
        batch_tensors = self.tokenizer(input_phrases, padding=True, truncation=True, return_tensors="pt")

        # Move tensors to the same device as the model
        device = next(self.parameters()).device
        batch_tensors = {k: v.to(device) for k, v in batch_tensors.items()}

        with torch.no_grad():
            encoded = self.bert(**batch_tensors, return_dict=True)
            pooled_output = encoded.pooler_output  # [batch_size, hidden_size]

        x = self.dropout(pooled_output)
        x = F.relu(self.fc1(x))
        x = self.dropout(x)
        x = F.relu(self.fc2(x))
        x = self.dropout(x)
        x = self.out(x)

        return x
````

### Retriever training: minibatches

Our model learns on triples $(q, a^+, a^-)$:
* q - <b>q</b>uestion
* (a+) - correct <b>a</b>nswer
* (a-) - wrong <b>a</b>nswer

Below you will find a generator that samples such triples from data.

````python
import random

def find_hard_negatives(question, options, wrong_indices, question_vectorizer, answer_vectorizer):
    """
    Find the hard negative - the wrong answer with highest similarity to the question

    :param question: The question string
    :param options: List of all possible answers
    :param wrong_indices: Indices of wrong answers in options
    :param question_vectorizer: Model to vectorize questions
    :param answer_vectorizer: Model to vectorize answers
    :returns: Index of the hard negative in options
    """
    if not wrong_indices:
        return None  # No wrong answers available

    # If models are not initialized yet, fall back to random selection
    if question_vectorizer is None or answer_vectorizer is None:
        return random.choice(wrong_indices)

    # Vectorize the question
    with torch.no_grad():
        q_vector = question_vectorizer([question])

        # Get all wrong answers
        wrong_answers = [options[i] for i in wrong_indices]

        # Vectorize all wrong answers
        wrong_vectors = answer_vectorizer(wrong_answers)

        # Calculate similarity between question and each wrong answer
        similarities = torch.sum(q_vector * wrong_vectors, dim=1)

        # Find the index of the wrong answer with highest similarity
        hard_negative_idx = torch.argmax(similarities).item()

        # Return the original index in options
        return wrong_indices[hard_negative_idx]

def iterate_minibatches(data, batch_size, shuffle=True, cycle=False, use_hard_negatives=False, 
                        question_vectorizer=None, answer_vectorizer=None):
    """
    Generates minibatches of triples: {questions, correct answers, wrong answers}
    If there are several wrong (or correct) answers, picks one at random.
    If use_hard_negatives is True, selects the wrong answer with highest similarity to the question.

    :param data: DataFrame with questions, options, correct_indices, wrong_indices
    :param batch_size: Size of each batch
    :param shuffle: Whether to shuffle the data
    :param cycle: Whether to cycle through the data indefinitely
    :param use_hard_negatives: Whether to use hard negative sampling
    :param question_vectorizer: Model to vectorize questions (needed for hard negatives)
    :param answer_vectorizer: Model to vectorize answers (needed for hard negatives)
    """
    indices = np.arange(len(data))
    while True:
        if shuffle:
            indices = np.random.permutation(indices)
        for batch_start in range(0, len(indices), batch_size):
            batch_indices = indices[batch_start: batch_start + batch_size]
            batch = data.iloc[batch_indices]
            questions = batch['question'].values

            correct_answers = np.array([
                row['options'][random.choice(row['correct_indices'])]
                for i, row in batch.iterrows()
            ])

            if use_hard_negatives and question_vectorizer is not None and answer_vectorizer is not None:
                # Put models in eval mode for inference
                question_vectorizer.eval()
                answer_vectorizer.eval()

                wrong_answers = np.array([
                    row['options'][find_hard_negatives(
                        row['question'], 
                        row['options'], 
                        row['wrong_indices'],
                        question_vectorizer,
                        answer_vectorizer
                    )]
                    for i, row in batch.iterrows()
                ])

                # Put models back in training mode
                question_vectorizer.train()
                answer_vectorizer.train()
            else:
                # Fall back to random selection if hard negatives can't be used
                wrong_answers = np.array([
                    row['options'][random.choice(row['wrong_indices'])]
                    for i, row in batch.iterrows()
                ])

            yield {
                'questions' : questions,
                'correct_answers': correct_answers,
                'wrong_answers': wrong_answers,
            }
        if not cycle:
            break
````

### Retriever training: loss function (2 points)
We want our vectorizers to put correct answers closer to question vectors and incorrect answers farther away from them. One way to express this is to use is Pairwise Hinge Loss _(aka Triplet Loss)_.

$$ L = \frac 1N \underset {q, a^+, a^-} \sum max(0, \space \delta - sim[V_q(q), V_a(a^+)] + sim[V_q(q), V_a(a^-)] )$$

, where
* sim[a, b] is some similarity function: dot product, cosine or negative distance
* δ - loss hyperparameter, e.g. δ=1.0. If sim[a, b] is linear in b, all δ > 0 are equivalent.


This reads as __Correct answers must be closer than the wrong ones by at least δ.__

![img](https://raw.githubusercontent.com/yandexdataschool/nlp_course/master/resources/margin.png)
<center>_image: question vector is green, correct answers are blue, incorrect answers are red_</center>


Note: in effect, we train a Deep Semantic Similarity Model [DSSM](https://www.microsoft.com/en-us/research/project/dssm/).

````python
def similarity(a, b):
    """ Dot product as a similarity function """
    return torch.sum(a * b, dim=1)

def compute_loss(question_vectors, correct_answer_vectors, wrong_answer_vectors, delta=1.0):
    """
    Compute the triplet loss as per formula above.
    Use similarity function above for  sim[a, b]
    :param question_vectors: float32[batch_size, vector_size]
    :param correct_answer_vectors: float32[batch_size, vector_size]
    :param wrong_answer_vectors: float32[batch_size, vector_size]
    :returns: loss for every row in batch, float32[batch_size]
    Hint: you can compute max(0, *) using torch.relu :)
    """
    sim_correct = similarity(question_vectors, correct_answer_vectors)
    sim_wrong = similarity(question_vectors, wrong_answer_vectors)

    # Triplet loss: max(0, delta - sim[q, c] + sim[q, w])
    loss = torch.relu(delta - sim_correct + sim_wrong)

    return loss
````

````python
dummy_v1 = torch.tensor([[0.1, 0.2, -1], [-1.2, 0.6, 1.0]], dtype=torch.float32)
dummy_v2 = torch.tensor([[0.9, 2.1, -6.6], [0.1, 0.8, -2.2]], dtype=torch.float32)
dummy_v3 = torch.tensor([[-4.1, 0.1, 1.2], [0.3, -1, -2]], dtype=torch.float32)

assert np.allclose(similarity(dummy_v1, dummy_v2).data.numpy(), [7.11, -1.84])
assert np.allclose(compute_loss(dummy_v1, dummy_v2, dummy_v3, delta=5.0).data.numpy(), [0.0, 3.88])
````

Once loss is working, let's train our model by our usual means.

````python
# we also compute recall: the rate at which a^+ is closer to q than a^-
def get_recall(questions, correct_answers, wrong_answers):
    # Convert numpy arrays to lists
    questions = questions.tolist() if isinstance(questions, np.ndarray) else questions
    correct_answers = correct_answers.tolist() if isinstance(correct_answers, np.ndarray) else correct_answers
    wrong_answers = wrong_answers.tolist() if isinstance(wrong_answers, np.ndarray) else wrong_answers

    # Vectorize questions and answers using our models
    v_questions = question_vectorizer(questions)
    v_correct = answer_vectorizer(correct_answers)
    v_incorrect = answer_vectorizer(wrong_answers)

    # Compute similarity between questions and answers
    correct_is_closer = similarity(v_questions, v_correct) > similarity(v_questions, v_incorrect)

    # Calculate recall as the mean of correct_is_closer
    recall = torch.mean(correct_is_closer.to(torch.float32)).item()

    return recall
````

### Training loop (1 point)

Just as we always do, we can now train DSSM on minibatches and periodically measure recall on validation data.


__Note 1:__ triplet loss training may be very sensitive to the choice of batch size. Small batch size may decrease model quality because there are less negative to consider.

__Note 2:__ here we use the same dataset as __"test set"__ and __"validation (dev) set"__.

In any serious scientific experiment, those must be two separate sets. Validation is for hyperparameter tuning and test is for final eval only.


````python
class DSSM(nn.Module):
    def __init__(self, hid_size=256):
        super().__init__()
        self.question_vectorizer = Vectorizer(hid_size=hid_size)
        self.answer_vectorizer = Vectorizer(hid_size=hid_size)

    def forward(self, questions, answers=None):
        """
        Forward pass of the model
        :param questions: list of question strings
        :param answers: list of answer strings (optional)
        :returns: question vectors and answer vectors if answers provided
        """
        # Convert numpy arrays to lists if needed
        questions = questions.tolist() if isinstance(questions, np.ndarray) else questions

        # Get question vectors
        question_vectors = self.question_vectorizer(questions)

        if answers is None:
            return question_vectors

        # Convert answer arrays to lists if needed
        answers = answers.tolist() if isinstance(answers, np.ndarray) else answers

        # Get answer vectors
        answer_vectors = self.answer_vectorizer(answers)

        return question_vectors, answer_vectors
````

### Hard Negatives Implementation

Hard negative sampling is a technique to improve the efficiency and effectiveness of contrastive learning. Instead of randomly selecting negative examples, we specifically choose the most challenging ones - those that are most similar to the query but are actually incorrect.

#### Why Hard Negatives Matter

1. **Training Efficiency**: As training progresses, random negatives become too easy to distinguish from positives, resulting in zero gradients (when using hinge/triplet loss). Hard negatives ensure the model continues to learn from challenging examples.

2. **Better Discrimination**: By focusing on examples that are most confusing to the model, hard negatives help the model learn more fine-grained distinctions between correct and incorrect answers.

3. **Faster Convergence**: Training with hard negatives often leads to faster convergence and better final performance.

#### Implementation Details

Our implementation includes:

1. **Hard Negative Mining**: We find the wrong answer with the highest similarity to the question:
   ```
   a^-_{hard} = argmax_{a^-} sim[V_q(q), V_a(a^-)]
   ```

2. **Batch Processing**: For each question in a batch, we:
   - Vectorize the question
   - Vectorize all wrong answers
   - Calculate similarity between the question and each wrong answer
   - Select the wrong answer with the highest similarity

3. **Training Integration**: We modified the training loop to use hard negatives by passing the model's vectorizers to the batch generator.

#### Comparing Hard Negatives vs Random Negatives

To demonstrate the effectiveness of hard negative sampling, we can compare the training progress with and without hard negatives.
Hard negatives help the model focus on the most challenging examples, which can lead to faster convergence and better performance.


````python
def train_model_with_config(use_hard_neg=False, num_iterations=500, batch_size=256, lr=1e-3):
    """
    Train a model with or without hard negatives and return training metrics

    :param use_hard_neg: Whether to use hard negative sampling
    :param num_iterations: Number of training iterations
    :param batch_size: Batch size for training
    :param lr: Learning rate
    :returns: Dictionary with training metrics
    """
    # Initialize model and optimizer
    model = DSSM()
    model.cuda() if torch.cuda.is_available() else model.to(device)
    optimizer = optim.Adam(model.parameters(), lr=lr)

    # Create batch generators
    train_batches = iterate_minibatches(
        train, 
        batch_size=batch_size, 
        cycle=True, 
        use_hard_negatives=use_hard_neg,
        question_vectorizer=model.question_vectorizer if use_hard_neg else None,
        answer_vectorizer=model.answer_vectorizer if use_hard_neg else None
    )
    dev_batches = iterate_minibatches(test, batch_size=batch_size, cycle=True)

    # Training metrics
    loss_history = []
    dev_recall_history = []

    # Training loop
    for i in tqdm(range(num_iterations)):
        # Get batch
        batch = next(train_batches)

        # Training step
        model.train()
        questions = batch['questions'].tolist()
        correct_answers = batch['correct_answers'].tolist()
        wrong_answers = batch['wrong_answers'].tolist()

        q_emb = model.question_vectorizer(questions)
        pos_emb = model.answer_vectorizer(correct_answers)
        neg_emb = model.answer_vectorizer(wrong_answers)

        # Compute loss
        loss = compute_loss(q_emb, pos_emb, neg_emb).mean()

        # Backpropagation
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        # Record loss
        loss_history.append(loss.item())

        # Evaluate on dev set every 50 iterations
        if (i + 1) % 50 == 0:
            dev_batch = next(dev_batches)
            model.eval()
            with torch.no_grad():
                recall = get_recall(
                    dev_batch['questions'].tolist(), 
                    dev_batch['correct_answers'].tolist(), 
                    dev_batch['wrong_answers'].tolist()
                )
            dev_recall_history.append(recall)

    return {
        'loss_history': loss_history,
        'dev_recall_history': dev_recall_history,
        'final_recall': np.mean(dev_recall_history[-3:]) if dev_recall_history else 0,
        'model': model
    }
````

# Retriever evaluation (2 point)

Let's see how well does our model perform on actual question answering.

Given a question and a set of possible answers, pick answer with highest similarity to estimate accuracy.

````python
# Optional: a helper to get embeddings from the model
def encode(texts):
    # Convert to tensor, batch, and move to CUDA if needed
    # Assume texts is a list of strings (e.g., ['question?', 'answer1', 'answer2', ...])
    with torch.no_grad():
        inputs = tokenizer(texts)  # You need to define a tokenizer that matches your model
        # inputs = inputs.cuda()
        embeddings = model(inputs)
    return embeddings

def select_best_answer(question, possible_answers):
    """
    Predicts which answer best fits the question
    :param question: a single string containing a question
    :param possible_answers: a list of strings containing possible answers
    :returns: integer - the index of best answer in possible_answers
    """
    model.eval()

    # Encode question and answers
    question_emb = encode([question])  # Shape: (1, d)
    answers_emb = encode(possible_answers)  # Shape: (n, d)

    # Compute cosine similarity
    similarity = F.cosine_similarity(question_emb, answers_emb)

    # Get index of the most similar answer
    best_idx = torch.argmax(similarity).item()

    return best_idx
````

### Optional tasks for bonus points

There are many ways to improve our question answering model. Here's a bunch of things you can do to increase your understanding and get bonus points.

### 1.  Hard Negatives (2+ pts)

Not all wrong answers are equally wrong. As the training progresses, _most negative examples $a^-$ will be to easy._ So easy in fact, that loss function and gradients on such negatives is exactly __0.0__. To improve training efficiency, one can __mine hard negative samples__.

Given a list of answers,
* __Hard negative__ is the wrong answer with highest similarity with question,

$$a^-_{hard} = \underset {a^-} {argmax} \space sim[V_q(q), V_a(a^-)]$$

* __Semi-hard negative__ is the one with highest similarity _among wrong answers that are farther than positive one. This option is more useful if some wrong answers may actually be mislabelled correct answers.

* One can also __sample__ negatives proportionally to $$P(a^-_i) \sim e ^ {sim[V_q(q), V_a(a^-_i)]}$$


The task is to implement at least __hard negative__ sampling and apply it for model training.

### 3. Search engine (3+ pts)

Our basic model only selects answers from 2-5 available sentences in paragraph. You can extend it to search over __the whole dataset__. All sentences in all other paragraphs are viable answers.

The goal is to train such a model and use it to __quickly find top-10 answers from the whole set__.

* You can ask such model a question of your own making - to see which answers it can find among the entire training dataset or even the entire wikipedia.
* Searching for top-K neighbors is easier if you use specialized methods: [KD-Tree](https://scikit-learn.org/stable/modules/generated/sklearn.neighbors.KDTree.html) or [HNSW](https://github.com/nmslib/hnswlib).
* This task is much easier to train if you use hard or semi-hard negatives. You can even find hard negatives for one question from correct answers to other questions in batch - do so in-graph for maximum efficiency. See [1.] for more details.


````python
from sklearn.neighbors import KDTree
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# -- Assume all_sentences and retriever are defined --

# Encode all sentences
sentence_embeddings = retriever.encode(all_sentences)

# Build KDTree
kdtree = KDTree(sentence_embeddings)

# Search function
def search(question, top_k=10):
    question_embedding = retriever.encode([question])[0].reshape(1, -1)
    top_k = min(top_k, len(all_sentences))
    distances, indices = kdtree.query(question_embedding, k=top_k)

    results = []
    for idx, dist in zip(indices[0], distances[0]):
        results.append((all_sentences[idx], dist))

    return results

# Hard Negative Mining
def find_hard_negatives(question, correct_sentence, all_sentences, retriever, k=10):
    question_emb = retriever.encode([question])[0].reshape(1, -1)
    sentence_embs = retriever.encode(all_sentences)

    sims = cosine_similarity(question_emb, sentence_embs)[0]

    hard_candidates = []
    for sent, sim in zip(all_sentences, sims):
        if sent != correct_sentence:
            hard_candidates.append((sent, sim))

    hard_candidates.sort(key=lambda x: -x[1])
    return hard_candidates[:k]

# Example
query = "Tell me something about animals."
correct_sentence = "Sentence 1 about cats."

# Search
print("Search Results:")
for sent, dist in search(query, top_k=3):
    print(f"- {sent} (distance {dist:.4f})")

# Hard Negatives
print("\nHard Negatives:")
for sent, sim in find_hard_negatives(query, correct_sentence, all_sentences, retriever, k=3):
    print(f"- {sent} (similarity {sim:.4f})")
````
